## 1. My lane as an ML task (type)

**Task Type: Ranking**

**Why:** The goal is to order a set of content items for a user based on their predicted relevance. This aligns with the core function of a content recommendation or search system. I need to determine which item is most likely to engage a specific user and present it first, followed by the next most relevant, and so on.


## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

## 2. Target or proxy

**Target:** User engagement with a recommended content item.

**Proxy:** Since we don't have direct feedback on "engagement satisfaction," I will use **click-through (1 = clicked, 0 = did not click)** as a proxy. A click indicates the user found the item sufficiently interesting to interact with it, which is a strong signal of relevance.

**Where the label comes from:** The label is **observed** from historical user interaction logs — we know which items were shown to which users, and whether they clicked.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

## 3. Success metric

**Primary Metric:** **Mean Reciprocal Rank (MRR)**

**Why:** MRR is a standard metric for ranking tasks. It measures how high the first relevant result appears in the ranked list. This is perfect because we want to ensure the most relevant item is at the very top of the list for the user.

**What number means 'good':** An MRR > 0.5 would be strong, meaning the first relevant item is typically in the top 2 positions. A baseline random ranking would give an MRR close to 0.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## 5. Why ML beats a fixed rule here

A fixed rule like "always recommend the most popular item" fails because:

1. **It's Not Personal:** It ignores individual user preferences. A user interested in niche topics will be shown generic, popular content they may not care about.

2. **It's Static:** Popularity changes over time. A rule can't adapt to new trends or shifts in user interest without manual updates.

3. **It's Simplistic:** User behavior depends on many complex factors (time of day, recent activity, item metadata, user history). An ML model can learn these intricate patterns automatically.

4. **It Can't Scale:** There are too many possible user-item combinations to manually encode rules for each scenario.

**Why ML:** Machine learning can learn patterns from multiple features simultaneously, adapt to new data, and make personalized predictions at scale — something a fixed if-statement cannot do.
